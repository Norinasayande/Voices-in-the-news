```python
import pandas as pd

# Load file
df = pd.read_csv("Sample - sample.csv")

# Grab country columns
predicted = df["primary_country"].str.strip().str.lower()
labeled = df["human_checked_country"].str.strip().str.lower()

# Compare
matches = predicted == labeled
accuracy = matches.mean()

print(f"Accuracy: {accuracy:.2%} ({matches.sum()} / {len(matches)} correct)\n")

# Show mismatches
mismatches = pd.DataFrame({
    "index": matches[~matches].index,
    "predicted": predicted[~matches].values,
    "labeled": labeled[~matches].values
})

print(f"Mismatches ({len(mismatches)}):")
print(mismatches.to_string(index=False))
```
